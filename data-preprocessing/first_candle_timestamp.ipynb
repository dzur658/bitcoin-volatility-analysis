The purpose of this notebook is to definitively declare when the first candle occurs for every coin.

```python
import os
import pandas as pd
```

```python
df_target_coins = pd.DataFrame(columns=["symbol", "first_candle"])

for file in os.listdir("unnormalized-feature-engineering"):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join("unnormalized-feature-engineering", file))
        df.sort_values("open_time", inplace=True)
        first_candle = df["open_time"].iloc[0]
        symbol = df["symbol"].iloc[0]
        df_target_coins = pd.concat([df_target_coins, pd.DataFrame({"symbol": symbol, "first_candle": first_candle}, index=[0])], ignore_index=True)
```

```python
print(df_target_coins.sort_values("first_candle"))
```

We chose the beginning of the day, of 2017-12-12 to open market-relative features. This provides 22 days for NEO to join `mkt_rvol_21`, and provides a stable cross section of coins to begin tracking the market those four starting coins are:

- ETH: dominant coin
